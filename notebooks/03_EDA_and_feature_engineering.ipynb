```python
# ------------------------------------------------------------
# STEP 2 — Row-level validity checks
# ------------------------------------------------------------
print("Rows:", len(parsed_df))
print("Columns:", len(parsed_df.columns))

# 1. Listing type distribution
print("\nlisting_type distribution:")
display(
    parsed_df["listing_type"]
    .value_counts(dropna=False)
    .to_frame("count")
    .assign(pct=lambda x: (x["count"] / len(parsed_df) * 100).round(2))
)

# 2. Training eligibility distribution
print("\ntraining_eligible distribution:")
display(
    parsed_df["training_eligible"]
    .value_counts(dropna=False)
    .to_frame("count")
    .assign(pct=lambda x: (x["count"] / len(parsed_df) * 100).round(2))
)

# 3. Deal type distribution
print("\ndeal_raw distribution:")
display(
    parsed_df["deal_raw"]
    .value_counts(dropna=False)
    .to_frame("count")
    .assign(pct=lambda x: (x["count"] / len(parsed_df) * 100).round(2))
)

# 4. Currency distribution
print("\ncurrency distribution:")
display(
    parsed_df["currency"]
    .value_counts(dropna=False)
    .to_frame("count")
    .assign(pct=lambda x: (x["count"] / len(parsed_df) * 100).round(2))
)

# 5. VAT status distribution
print("\nvat_status distribution:")
display(
    parsed_df["vat_status"]
    .value_counts(dropna=False)
    .to_frame("count")
    .assign(pct=lambda x: (x["count"] / len(parsed_df) * 100).round(2))
)

# 6. Duplicate checks
duplicate_summary = {
    "duplicate_ad_url_rows": parsed_df.duplicated("ad_url").sum(),
    "duplicate_ad_id_rows": parsed_df.duplicated("ad_id").sum(),
    "duplicate_title_rows": parsed_df.duplicated("title").sum(),
    "duplicate_html_path_rows": parsed_df.duplicated("html_path").sum(),
}

print("\nDuplicate summary:")
display(pd.DataFrame([duplicate_summary]))

# 7. Inspect duplicated ad_id rows
dup_ad_ids = parsed_df.loc[parsed_df.duplicated("ad_id", keep=False), [
    "ad_id",
    "ad_url",
    "title",
    "total_price",
    "area_sqm",
    "price_per_sqm",
    "location_raw",
]]

print("\nDuplicated ad_id rows:")
display(dup_ad_ids.sort_values("ad_id"))

# 8. Core ML validity flags
validity = pd.DataFrame({
    "has_deal": parsed_df["deal_raw"].notna(),
    "has_property_type": parsed_df["property_type_raw"].notna(),
    "has_city": parsed_df["title_city_raw"].notna(),
    "has_area": parsed_df["area_sqm"].notna(),
    "has_total_price": parsed_df["total_price"].notna(),
    "has_price_per_sqm": parsed_df["price_per_sqm"].notna(),
    "positive_area": parsed_df["area_sqm"].gt(0),
    "positive_total_price": parsed_df["total_price"].gt(0),
    "positive_price_per_sqm": parsed_df["price_per_sqm"].gt(0),
    "training_eligible": parsed_df["training_eligible"].eq(True),
})

validity_summary = (
    validity
    .sum()
    .to_frame("valid_count")
    .assign(
        invalid_count=lambda x: len(parsed_df) - x["valid_count"],
        valid_pct=lambda x: (x["valid_count"] / len(parsed_df) * 100).round(2)
    )
)

print("\nValidity summary:")
display(validity_summary)

# 9. First strict candidate rows for regression
df_candidate = parsed_df[
    parsed_df["training_eligible"].eq(True)
    & parsed_df["deal_raw"].notna()
    & parsed_df["property_type_raw"].notna()
    & parsed_df["title_city_raw"].notna()
    & parsed_df["area_sqm"].gt(0)
    & parsed_df["total_price"].gt(0)
    & parsed_df["price_per_sqm"].gt(0)
].copy()

print("\nFirst-pass ML candidate rows:", len(df_candidate))
print("Share of dataset:", round(len(df_candidate) / len(parsed_df) * 100, 2), "%")

# 10. Candidate property type distribution
print("\nCandidate property_type_raw distribution:")
display(
    df_candidate["property_type_raw"]
    .value_counts(dropna=False)
    .to_frame("count")
    .assign(pct=lambda x: (x["count"] / len(df_candidate) * 100).round(2))
    .head(50)
)
```

```python
MIN_AREA_BY_PROPERTY_TYPE = {
    "1-СТАЕН": 15,
    "2-СТАЕН": 25,
    "3-СТАЕН": 35,
    "4-СТАЕН": 45,
    "МНОГОСТАЕН": 50,
    "МЕЗОНЕТ": 40,
    "АТЕЛИЕ, ТАВАН": 10,
    "КЪЩА": 25,
    "ЕТАЖ ОТ КЪЩА": 25,
    "ВИЛА": 20,
    "ГАРАЖ": 8,
    "ПАРКОМЯСТО": 8,
    "МАГАЗИН": 5,
    "ОФИС": 5,
    "СКЛАД": 5,
    "ПАРЦЕЛ": 50,
    "ЗЕМЕДЕЛСКА ЗЕМЯ": 100,
}

df_ml_base["min_reasonable_area"] = (
    df_ml_base["property_type_raw"]
    .map(MIN_AREA_BY_PROPERTY_TYPE)
)

df_ml_base["suspicious_area_too_small"] = (
    df_ml_base["min_reasonable_area"].notna()
    & (df_ml_base["area_sqm"] < df_ml_base["min_reasonable_area"])
)

small_area_review = df_ml_base[
    df_ml_base["suspicious_area_too_small"]
][[
    "ad_id",
    "ad_url",
    "property_type_raw",
    "title_city_raw",
    "location_raw",
    "total_price",
    "area_sqm",
    "price_per_sqm",
    "title",
]].sort_values(["property_type_raw", "area_sqm"])

print("Suspicious too-small area rows:", len(small_area_review))
display(small_area_review.head(50))
```

```python
# ------------------------------------------------------------
# Step 2B: Further filter the 338 suspicious area candidates
# using price_per_sqm
# ------------------------------------------------------------

# Replace this with the actual name of your 338-row dataframe if different
area_suspicious_df = small_area_review.copy()

print("Starting suspicious area candidates:", len(area_suspicious_df))


# ------------------------------------------------------------
# 1. Add a practical top price-per-sqm flag
# ------------------------------------------------------------

# These are review thresholds, not final ML rules.
# Keep them intentionally simple for now.

area_suspicious_df["very_high_price_per_sqm"] = (
    area_suspicious_df["price_per_sqm"] >= 10_000
)

area_suspicious_df["extreme_price_per_sqm"] = (
    area_suspicious_df["price_per_sqm"] >= 25_000
)

area_suspicious_df["absurd_price_per_sqm"] = (
    area_suspicious_df["price_per_sqm"] >= 50_000
)


# ------------------------------------------------------------
# 2. Add combined drop candidate logic
# ------------------------------------------------------------

area_suspicious_df["likely_bad_area_or_price"] = (
    (area_suspicious_df["area_sqm"] <= 2)
    | (
        (area_suspicious_df["area_sqm"] <= 10)
        & (area_suspicious_df["price_per_sqm"] >= 10_000)
    )
    | (area_suspicious_df["price_per_sqm"] >= 50_000)
)


# ------------------------------------------------------------
# 3. Inspect the strongest drop candidates first
# ------------------------------------------------------------

drop_review = area_suspicious_df[
    area_suspicious_df["likely_bad_area_or_price"]
][[
    "ad_id",
    "ad_url",
    "property_type_raw",
    "title_city_raw",
    "location_raw",
    "total_price",
    "area_sqm",
    "price_per_sqm",
    "title",
]].sort_values(
    ["price_per_sqm", "area_sqm"],
    ascending=[False, True]
)

print("Likely bad rows from the 338 candidates:", len(drop_review))
display(drop_review.head(100))


# ------------------------------------------------------------
# 4. Inspect borderline candidates separately
# ------------------------------------------------------------

borderline_review = area_suspicious_df[
    ~area_suspicious_df["likely_bad_area_or_price"]
][[
    "ad_id",
    "ad_url",
    "property_type_raw",
    "title_city_raw",
    "location_raw",
    "total_price",
    "area_sqm",
    "price_per_sqm",
    "title",
]].sort_values(
    ["area_sqm", "price_per_sqm"],
    ascending=[True, False]
)

print("Borderline rows still needing manual review:", len(borderline_review))
display(borderline_review.head(100))
```

```python
# ------------------------------------------------------------
# Step 2D: Detect likely decare-parsed-as-sqm issue
# ------------------------------------------------------------

LAND_DECARE_LIKELY_TYPES = {
    "ЗЕМЕДЕЛСКА ЗЕМЯ",
}

df_candidate = df_candidate.copy()

df_candidate["area_unit_suspected"] = None
df_candidate["area_sqm_corrected"] = df_candidate["area_sqm"]
df_candidate["price_per_sqm_corrected"] = df_candidate["price_per_sqm"]

likely_decare_mask = (
    df_candidate["property_type_raw"].isin(LAND_DECARE_LIKELY_TYPES)
    & df_candidate["area_sqm"].between(0.1, 100)
    & (df_candidate["price_per_sqm"] > 500)
)

df_candidate.loc[likely_decare_mask, "area_unit_suspected"] = "decare_parsed_as_sqm"

df_candidate.loc[likely_decare_mask, "area_sqm_corrected"] = (
    df_candidate.loc[likely_decare_mask, "area_sqm"] * 1000
)

df_candidate.loc[likely_decare_mask, "price_per_sqm_corrected"] = (
    df_candidate.loc[likely_decare_mask, "total_price"]
    / df_candidate.loc[likely_decare_mask, "area_sqm_corrected"]
)

print("Likely decare correction rows:", likely_decare_mask.sum())

display(
    df_candidate.loc[
        likely_decare_mask,
        [
            "ad_id",
            "property_type_raw",
            "title_city_raw",
            "location_raw",
            "total_price",
            "area_sqm",
            "area_sqm_corrected",
            "price_per_sqm",
            "price_per_sqm_corrected",
            "title",
        ],
    ]
    .sort_values("price_per_sqm", ascending=False)
    .head(100)
)
```

```python
# ------------------------------------------------------------
# Safer geo category mapping
# Fix false positives like "град Търговище, Боровец"
# ------------------------------------------------------------

def map_geo_category(row):
    level_1_raw = "" if pd.isna(row.get("location_level_1_model")) else str(row.get("location_level_1_model")).strip().lower()
    level_2_raw = "" if pd.isna(row.get("location_level_2_model")) else str(row.get("location_level_2_model")).strip().lower()

    level_1 = row.get("location_level_1_norm", "")
    level_2 = row.get("location_level_2_norm", "")
    level_3 = row.get("location_level_3_norm", "")

    level_1 = "" if pd.isna(level_1) else str(level_1)
    level_2 = "" if pd.isna(level_2) else str(level_2)
    level_3 = "" if pd.isna(level_3) else str(level_3)

    # Sofia first, because Sofia has "Лозенец", which should not become sea_resort.
    if level_1_raw == "град софия":
        if level_2 in capital_center_areas:
            return "capital_center"
        return "capital_non_center"

    # Sea resort:
    # Allow direct city resort, or resort settlement inside oblast.
    if (
        level_1 in sea_resorts
        or (
            level_1_raw.startswith("област ")
            and level_2 in sea_resorts
        )
    ):
        return "sea_resort"

    # Mountain resort:
    # Context-aware to avoid false positives like "град Търговище, Боровец".
    if level_1 in mountain_resorts:
        return "mountain_resort"

    if level_1_raw.startswith("област ") and level_2 in mountain_resorts:
        return "mountain_resort"

    # Special case: Borovets should only be mountain resort if under Sofia/Samokov context.
    if level_2 == "боровец":
        if (
            level_1_raw == "област софия"
            or level_1 == "самоков"
        ):
            return "mountain_resort"

    if level_1 in large_regional_cities:
        return "large_regional_city"

    if level_1 in regional_cities:
        return "regional_city"

    if level_1_raw.startswith("област ") and (
        level_2_raw.startswith("с. ")
        or level_2_raw.startswith("село ")
    ):
        return "village"

    if level_1_raw.startswith("област ") and (
        level_2_raw.startswith("гр. ")
        or level_2_raw.startswith("град ")
        or level_2_raw.startswith("к.к. ")
    ):
        return "small_city"

    if level_1_raw.startswith("град "):
        return "small_city"

    return "other_unknown"


df_model_base["geo_category"] = df_model_base.apply(map_geo_category, axis=1)

print("Geo category distribution:")
display(
    df_model_base["geo_category"]
    .value_counts(dropna=False)
    .rename_axis("geo_category")
    .reset_index(name="rows")
)

print("\nMountain resort sample after fix:")
display(
    df_model_base.loc[
        df_model_base["geo_category"].eq("mountain_resort"),
        [
            "ad_id",
            "property_type_raw",
            "location_raw",
            "location_level_1_model",
            "location_level_2_model",
            "location_level_2_norm",
            "total_price",
            "area_sqm",
            "price_per_sqm",
        ]
    ].sample(
        min(30, (df_model_base["geo_category"] == "mountain_resort").sum()),
        random_state=42,
    )
)
```

```python
# ------------------------------------------------------------
# Better fix for remaining other_unknown using title_city_raw + title_geo_2_raw
# No blanket "oblast -> small_city"
# ------------------------------------------------------------

df_model_base = df_model_base.copy()

# Normalize title fields
df_model_base["title_city_model"] = (
    df_model_base["title_city_raw"]
    .astype("string")
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

df_model_base["title_geo_2_model"] = (
    df_model_base["title_geo_2_raw"]
    .astype("string")
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
)

# Drop Greece based on title_city_raw
before_rows = len(df_model_base)

drop_greece_mask = df_model_base["title_city_model"].eq("гърция")

dropped_greece_title_city = df_model_base.loc[drop_greece_mask].copy()
df_model_base = df_model_base.loc[~drop_greece_mask].copy()

print("Dropped Greece rows:", len(dropped_greece_title_city))


large_regional_city_titles = {
    "град пловдив",
    "град варна",
    "град бургас",
}

regional_city_titles = {
    "град русе",
    "град стара загора",
    "град плевен",
    "град велико търново",
    "град шумен",
    "град добрич",
    "град сливен",
    "град хасково",
    "град ямбол",
    "град перник",
    "град благоевград",
    "град пазарджик",
    "град враца",
    "град габрово",
    "град кърджали",
    "град кюстендил",
    "град ловеч",
    "град монтана",
    "град разград",
    "град силистра",
    "град смолян",
    "град търговище",
    "град видин",
}

capital_center_areas_from_title = {
    "център",
    "идеален център",
    "докторски паметник",
    "яворов",
    "оборише",
    "оборище",
    "лозенец",
    "иван вазов",
    "зона б-5",
    "зона б-18",
    "зона б-19",
    "банишора",
    "сердика",
    "възраждане",
    "медицинска академия",
    "редута",
}


def map_unknown_from_title_city_geo(row):
    title_city = row.get("title_city_model")
    title_geo_2 = row.get("title_geo_2_model")

    title_city = "" if pd.isna(title_city) else str(title_city).strip().lower()
    title_geo_2 = "" if pd.isna(title_geo_2) else str(title_geo_2).strip().lower()

    # Sofia: use title_geo_2 to split center vs non-center
    if title_city == "град софия":
        if title_geo_2 in capital_center_areas_from_title:
            return "capital_center"
        return "capital_non_center"

    # Villages under oblast
    if title_city.startswith("област ") and (
        title_geo_2.startswith("с. ")
        or title_geo_2.startswith("село ")
    ):
        return "village"

    # Towns under oblast
    if title_city.startswith("област ") and (
        title_geo_2.startswith("гр. ")
        or title_geo_2.startswith("град ")
        or title_geo_2.startswith("к.к. ")
    ):
        return "small_city"

    # Large regional city direct
    if title_city in large_regional_city_titles:
        return "large_regional_city"

    # Regional city direct
    if title_city in regional_city_titles:
        return "regional_city"

    # Other direct city
    if title_city.startswith("град "):
        return "small_city"

    # Region without usable settlement detail
    if title_city.startswith("област "):
        return "small_city"

    return "other_unknown"


# Reclassify only remaining other_unknown
remaining_unknown_mask = df_model_base["geo_category"].eq("other_unknown")

reclassified_from_title_fields = df_model_base.loc[remaining_unknown_mask].copy()

df_model_base.loc[remaining_unknown_mask, "geo_category"] = (
    df_model_base.loc[remaining_unknown_mask]
    .apply(map_unknown_from_title_city_geo, axis=1)
)

print("Rows before:", before_rows)
print("Rows after:", len(df_model_base))
print("Rows reclassified from title fields:", len(reclassified_from_title_fields))

print("\nUpdated geo_category distribution:")
display(
    df_model_base["geo_category"]
    .value_counts(dropna=False)
    .rename_axis("geo_category")
    .reset_index(name="rows")
)

remaining_other_unknown = df_model_base.loc[
    df_model_base["geo_category"].eq("other_unknown")
].copy()

print("\nRemaining other_unknown rows:", len(remaining_other_unknown))

display(
    remaining_other_unknown[
        [
            "ad_id",
            "ad_url",
            "property_type_raw",
            "title_city_raw",
            "title_geo_2_raw",
            "location_raw",
            "title_city_model",
            "title_geo_2_model",
            "total_price",
            "area_sqm",
            "price_per_sqm",
        ]
    ]
    .head(100)
)

cleanup_actions.append({
    "rule": "drop_greece_by_title_city",
    "reason": "Greece listings are outside the Bulgarian target market; title_city_raw is the reliable field when location_raw is corrupted",
    "rows_affected": len(dropped_greece_title_city),
})

cleanup_actions.append({
    "rule": "reclassify_other_unknown_using_title_city_and_geo_2",
    "reason": "Rows with corrupted location_raw were reclassified using title_city_raw and title_geo_2_raw, preserving village vs small-city distinction",
    "rows_affected": len(reclassified_from_title_fields),
})
```

```python
# ------------------------------------------------------------
# Normalize title_city_raw and title_geo_2_raw
# - lowercase
# - trim whitespace
# - remove settlement prefixes from title_geo_2
# ------------------------------------------------------------

df_model_base = df_model_base.copy()


def normalize_bg_text(value):
    if pd.isna(value):
        return pd.NA

    value = str(value).strip().lower()
    value = value.replace("ё", "е")
    value = value.replace("̀", "")
    value = value.replace("'", "")
    value = value.replace('"', "")
    value = " ".join(value.split())

    if value in {"", "none", "nan", "<na>"}:
        return pd.NA

    return value


def remove_geo_prefix(value):
    if pd.isna(value):
        return pd.NA

    value = normalize_bg_text(value)

    if pd.isna(value):
        return pd.NA

    prefixes = [
        "гр. ",
        "град ",
        "с. ",
        "село ",
        "к.к. ",
        "кк ",
        "курорт ",
        "ж.к. ",
        "кв. ",
        "м-т ",
        "м. ",
        "местност ",
        "в.з. ",
        "с.о. ",
        "п.к. ",
    ]

    changed = True
    while changed:
        changed = False

        for prefix in prefixes:
            if value.startswith(prefix):
                value = value[len(prefix):].strip()
                changed = True

    return value if value else pd.NA


# Normalized city field
df_model_base["title_city_model"] = (
    df_model_base["title_city_raw"]
    .apply(normalize_bg_text)
)

# Normalized geo_2 field, with prefix removed
df_model_base["title_geo_2_model"] = (
    df_model_base["title_geo_2_raw"]
    .apply(remove_geo_prefix)
)


print("Unique title_city_raw:", df_model_base["title_city_raw"].nunique(dropna=False))
print("Unique title_city_model:", df_model_base["title_city_model"].nunique(dropna=False))

print("Unique title_geo_2_raw:", df_model_base["title_geo_2_raw"].nunique(dropna=False))
print("Unique title_geo_2_model:", df_model_base["title_geo_2_model"].nunique(dropna=False))


print("\nTop title_city_model:")
display(
    df_model_base["title_city_model"]
    .value_counts(dropna=False)
    .head(50)
    .rename_axis("title_city_model")
    .reset_index(name="rows")
)

print("\nTop title_geo_2_model:")
display(
    df_model_base["title_geo_2_model"]
    .value_counts(dropna=False)
    .head(100)
    .rename_axis("title_geo_2_model")
    .reset_index(name="rows")
)


cleanup_actions.append({
    "rule": "normalize_title_city_and_geo_2",
    "reason": "Created lowercase normalized location fields and removed settlement/location prefixes from title_geo_2 for cleaner geo mapping",
    "rows_affected": 0,
    "columns_added": 2,
})
```

```python
# ------------------------------------------------------------
# Final cleanup: create ML-ready dataset and save to Parquet
# ------------------------------------------------------------

# Choose your output path
ML_OUTPUT_PATH = PARSED_DATA_DIR / "imot_ml_ready.parquet"

df_ml_ready = df_model_base.copy()

# ------------------------------------------------------------
# 1. Keep only model-useful columns
# ------------------------------------------------------------

ml_ready_columns = [
    # Traceability - keep for audit, drop before fitting if needed
    "ad_id",
    "ad_url",

    # Core categorical features
    "listing_type",
    "property_type_raw",
    "currency",
    "vat_status",
    "title_city_model",
    "title_geo_2_model",
    "geo_category",

    # Numeric target / features
    "total_price",
    "area_sqm",
    "price_per_sqm",
    "views",
    "features_count",

    # Cleaned construction features
    "construction_type_model",
    "construction_year_model",

    # Cleaned floor features
    "floor_model",
    "total_floors_model",
    "floor_applicability",

    # Feature engineering input
    "features_pipe",
]

ml_ready_columns = [
    col for col in ml_ready_columns
    if col in df_ml_ready.columns
]

df_ml_ready = df_ml_ready[ml_ready_columns].copy()


# ------------------------------------------------------------
# 2. Standardize missing categorical values
# ------------------------------------------------------------

categorical_cols = [
    "listing_type",
    "property_type_raw",
    "currency",
    "vat_status",
    "title_city_model",
    "title_geo_2_model",
    "geo_category",
    "construction_type_model",
    "floor_applicability",
    "features_pipe",
]

for col in categorical_cols:
    if col in df_ml_ready.columns:
        df_ml_ready[col] = (
            df_ml_ready[col]
            .astype("string")
            .str.strip()
        )
        df_ml_ready[col] = df_ml_ready[col].fillna("unknown")


# ------------------------------------------------------------
# 3. Ensure numeric columns are numeric
# ------------------------------------------------------------

numeric_cols = [
    "total_price",
    "area_sqm",
    "price_per_sqm",
    "views",
    "features_count",
    "construction_year_model",
    "floor_model",
    "total_floors_model",
]

for col in numeric_cols:
    if col in df_ml_ready.columns:
        df_ml_ready[col] = pd.to_numeric(df_ml_ready[col], errors="coerce")


# ------------------------------------------------------------
# 4. Final sanity checks
# ------------------------------------------------------------

print("ML-ready rows:", len(df_ml_ready))
print("ML-ready columns:", df_ml_ready.shape[1])

print("\nColumns:")
print(df_ml_ready.columns.tolist())

print("\nMissing values:")
display(
    df_ml_ready
    .isna()
    .sum()
    .rename("missing")
    .reset_index()
    .rename(columns={"index": "column"})
    .sort_values("missing", ascending=False)
)

print("\nInfo:")
display(df_ml_ready.info())


# ------------------------------------------------------------
# 5. Save to Parquet
# ------------------------------------------------------------

ML_OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

df_ml_ready.to_parquet(
    ML_OUTPUT_PATH,
    index=False,
)

print(f"\nSaved ML-ready dataset to: {ML_OUTPUT_PATH}")
print("Saved rows:", len(df_ml_ready))
print("Saved columns:", df_ml_ready.shape[1])


# ------------------------------------------------------------
# 6. Optional: verify reload
# ------------------------------------------------------------

df_ml_ready_check = pd.read_parquet(ML_OUTPUT_PATH)

print("\nReload check:")
print("Rows:", len(df_ml_ready_check))
print("Columns:", df_ml_ready_check.shape[1])

display(df_ml_ready_check.head())
```
