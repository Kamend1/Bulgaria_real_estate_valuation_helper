```python
#Import Libraries here

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
from pathlib import Path
from IPython.display import display
```

```python
#Import your modules here
PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))
PARSED_DATA_DIR = PROJECT_ROOT / "data" / "parsed_sales_full"
```

```python
ml_model_base_df = pd.read_parquet(PARSED_DATA_DIR / "imot_ml_ready.parquet")
ml_model_base_df
```

```python
ml_model_base_df.title_city_model.value_counts()
```

```python
model_sofia = ml_model_base_df[ml_model_base_df.title_city_model == "град софия"]
model_sofia.info()
```

```python
model_sofia.geo_category.value_counts()
```

```python
model_sofia_non_center = model_sofia[model_sofia.geo_category == 'capital_non_center']
model_sofia_non_center.info()
```

```python
model_sofia_non_center.property_type_raw.value_counts()
```

```python
model_sofia_non_center_office = model_sofia_non_center[model_sofia_non_center.property_type_raw == 'ОФИС']
model_sofia_non_center_office
```

```python
model_sofia_non_center_office.price_per_sqm.describe()
```

```python
office_types = [
    'ЗАВЕДЕНИЕ',
    'МАГАЗИН',
    'ТЪРГОВСКИ КОМПЛЕКС',
    'СПА/СОЛАРНО СТУДИО',
    'АПТЕКА',
    'БАНКОВ ОФИС',
    'ФИТНЕС ЗАЛА',
    'ДЕТСКИ ЦЕНТЪР',
    'ФРИЗЬОРСКИ/КОЗМЕТИЧЕН САЛОН',
    'ЛЕКАРСКИ КАБИНЕТ',
    'СТОМАТОЛОГИЧЕН КАБИНЕТ'
]

model_sofia_non_center_shops = model_sofia_non_center[
    model_sofia_non_center['property_type_raw'].isin(office_types)
]

model_sofia_non_center_shops.info()
```

```python
model_sofia_non_center_shops.price_per_sqm.describe()
```

```python
parking_types = [
    'ПАРКОМЯСТО',

]

model_sofia_non_center_parking = model_sofia_non_center[
    model_sofia_non_center['property_type_raw'].isin(parking_types)
]

model_sofia_non_center_parking.info()
```

```python
model_sofia_non_center_parking.price_per_sqm.describe()
```

```python
output_path = r"D:\Users\kamen.dimitrov\Desktop\sofia_non_center_market_data.xlsx"

with pd.ExcelWriter(output_path, engine="xlsxwriter") as writer:
    model_sofia_non_center_shops.to_excel(writer, sheet_name="Shops", index=False)
    model_sofia_non_center_office.to_excel(writer, sheet_name="Office", index=False)
    model_sofia_non_center_parking.to_excel(writer, sheet_name="Parking", index=False)
```
